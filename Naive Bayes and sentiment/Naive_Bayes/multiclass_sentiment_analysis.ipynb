## Text Cleaning¶
1.Sample 10000 rows

2.Remove html tags

3.Remove special characters

4.Converting every thing to lower case

5.Removing Stop words

6.Stemming

````python
import re 
def clean_html(text):
    clean = re.compile('<.*?>')
    return re.sub(clean, '', text)
````

````python
df['text']=df['text'].apply(clean_html)
````

````python
df.head(5)
````

````python
# converting everything to lower

def convert_lower(text):
    return text.lower()
````

````python
df['text']=df['text'].apply(convert_lower)
````

````python
df.head(5)
````

````python
# function to remove special characters

def remove_special(text):
    x=''
    
    for i in text:
        if i.isalnum():
            x=x+i
        else:
            x=x + ' '
    return x
````

````python
df['text']=df['text'].apply(remove_special)
````

````python
df.head(5)
````

````python
# Remove the stop words
import nltk
````

````python
from nltk.corpus import stopwords
````

````python
stopwords.words('english')
````

````python
def remove_stopwords(text):
    x=[]
    for i in text.split():
        
        if i not in stopwords.words('english'):
            x.append(i)
    y=x[:]
    x.clear()
    return y
````

````python
df['text']=df['text'].apply(remove_stopwords)
````

````python
df
````

````python
# Perform stemming

from nltk.stem.porter import PorterStemmer
ps=PorterStemmer()
````

````python

def stem_words(text):
    y=[]
    for i in text:
        y.append(ps.stem(i))
    z=y[:]
    y.clear()
    return z
````

````python
stem_words(['I','loved','loving','it'])
````

````python
df['text']=df['text'].apply(stem_words)
````

````python
df.head(5)
````

````python
# Join back

def join_back(list_input):
    return " ".join(list_input)
    
````

````python
df['text']=df['text'].apply(join_back)
````

##  Predict on New Text (bonus — Original label-এ ফেরত আনতে `l লাগবে)


````python
def predict_category(text, model=clf2):
    cleaned = join_back(stem_words(remove_stopwords(remove_special(convert_lower(clean_html(text))))))
    vec = cv.transform([cleaned]).toarray()
    pred_enc = model.predict(vec)
    return category_map[pred_enc[0]]

predict_category("The government passed a new tax bill in parliament today.")
````
